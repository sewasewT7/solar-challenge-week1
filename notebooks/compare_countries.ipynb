```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway, kruskal

# Set plot style
sns.set(style="whitegrid")
```

```python
benin = pd.read_csv('data/benin_clean.csv')
sierra_leone = pd.read_csv('data/sierra_leone_clean.csv')
togo = pd.read_csv('data/togo_clean.csv')

# Add country column for distinction
benin['Country'] = 'Benin'
sierra_leone['Country'] = 'Sierra Leone'
togo['Country'] = 'Togo'

# Combine datasets
df = pd.concat([benin, sierra_leone, togo], ignore_index=True)
```

```python
metrics = ['GHI', 'DNI', 'DHI']

for metric in metrics:
    plt.figure(figsize=(8, 5))
    sns.boxplot(x='Country', y=metric, data=df, palette="Set2")
    plt.title(f'{metric} Distribution by Country')
    plt.ylabel(metric)
    plt.xlabel('Country')
    plt.tight_layout()
    plt.show()
```

```python
summary_table = df.groupby('Country')[metrics].agg(['mean', 'median', 'std']).round(2)
summary_table
```

```python
ghi_benin = benin['GHI']
ghi_sierra = sierra_leone['GHI']
ghi_togo = togo['GHI']

# ANOVA
anova_result = f_oneway(ghi_benin, ghi_sierra, ghi_togo)
print(f"One-way ANOVA p-value for GHI: {anova_result.pvalue:.4f}")

# Kruskal–Wallis Test (if data not normally distributed)
kruskal_result = kruskal(ghi_benin, ghi_sierra, ghi_togo)
print(f"Kruskal–Wallis p-value for GHI: {kruskal_result.pvalue:.4f}")
```

### Key Observations

- Benin has the **highest average GHI**, but Sierra Leone shows **less variability** in GHI values.
- Togo exhibits **greater variability** in DNI values compared to the others.
- Statistical testing reveals that the **difference in GHI across countries is significant** (`p < 0.05`).


```python
avg_ghi = df.groupby('Country')['GHI'].mean().sort_values(ascending=False)

plt.figure(figsize=(6,4))
sns.barplot(x=avg_ghi.index, y=avg_ghi.values, palette="viridis")
plt.title("Average GHI by Country")
plt.ylabel("Average GHI")
plt.xlabel("Country")
plt.tight_layout()
plt.show()
```
